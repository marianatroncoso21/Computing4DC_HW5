# Hh5: Diabetes Analysis

## Classes and Data

```python
import sys
sys.path.append("../src")

from sklearn.metrics import roc_auc_score

from computing4dc_hw5.loading.loaders import CSVLoader
from computing4dc_hw5.preprocessing import DropNaRows, FillNaWithMean
from computing4dc_hw5.features import BMIFeature, GenderEncoder
from computing4dc_hw5.modeling.models import LogisticModel
```

```python
loader = CSVLoader("../data/sample_diabetes_mellitus_data.csv")
train, test = loader.load()
print("Train:", train.shape, "Test:", test.shape)
```

## Preprocess

```python
preprocessors = [
    DropNaRows(["age", "gender", "ethnicity"]),
    FillNaWithMean(["height", "weight"]),
]

for preprocessor in preprocessors:
    train = preprocessor.transform(train)
    test = preprocessor.transform(test)

print("Train:", train.shape, "Test:", test.shape)
```

## Features

```python
features = [BMIFeature(), GenderEncoder()]

for feature in features:
    train = feature.transform(train)
    test = feature.transform(test)

train[["height", "weight", "bmi_calc", "gender", "is_female"]]
```

## Model

```python
feature_cols = ["age", "height", "weight", "bmi_calc", "is_female"]
target_col = "diabetes_mellitus"

model = LogisticModel(feature_cols, target_col)
model.train(train)
```

## Prediction

```python
test["predictions"] = model.predict(test)

auc = roc_auc_score(test[target_col], test["predictions"])
print(f"ROC AUC on test: {auc:.3f}")
```
